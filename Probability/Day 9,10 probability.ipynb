# Probability

## Probability :define as possibilities of events

```python
favorable = 3
total = 5

probability = favorable/total
print(probability)
```

## Conditional probability

### Conditional Probability is the probability of an event happening when we already know that another event has happened
### P(A|B) = P(A ∩ B) / P(B)

```python
# Number of students who passed and studied
passed_and_studied = 40

# Number of students who studied
studied = 50

# Conditional Probability
probability = passed_and_studied / studied

print(probability)
```

## Bayes' Theorem

### Bayes' Theorem helps us update the probability of an event when we get new information/evidence.

### P(A|B) = P(B|A) × P(A) / P(B)

```python
# Probability that an email is spam
p_spam = 0.4

# Probability of seeing the word "winner" if email is spam
p_word_given_spam = 0.8

# Probability that any email contains the word "winner"
p_word = 0.5

# Bayes' Theorem
p_spam_given_word = (
    p_word_given_spam * p_spam
) / p_word

print(p_spam_given_word)
```

## Random variable

## Joint Distribution

### Joint Distribution shows the probability of two variables occurring together.

```python
import pandas as pd

df = pd.DataFrame({
    "Weather": ["Hot", "Hot", "Cold", "Cold"],
    "Sales": ["High", "Low", "Low", "High"]
})

# Joint distribution
joint = pd.crosstab(df["Weather"], df["Sales"], normalize=True)
print(joint)
```

```python
# Types of Distribution

# 1. Joint → Together
# P(X, Y)

# 2. Marginal → Single variable
# P(X) or P(Y)

# 3. Conditional → Given another variable
# P(X | Y)
```
