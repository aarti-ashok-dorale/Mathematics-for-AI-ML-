# Visualization

## 1. Histogram

```python
import matplotlib.pyplot as plt

marks = [40, 42, 45, 48, 50, 52, 55, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90, 200]

plt.hist(marks)
plt.xlabel("marks")
plt.ylabel("frequency")
plt.show()
```

## 2. Boxplot

```python
plt.boxplot(marks)
plt.ylabel("marks")

plt.show()
```

## 3. Barchart

```python
subjects = ['maths', 'English', 'python', 'java']
marks = [80, 90, 75, 89]

plt.bar(subjects, marks)
plt.show()
```

## 4.Scatter plot

```python
study_hours = [1,2,3,4,5,10]
marks = [40,50,60,70,80, 100]

plt.scatter(study_hours, marks)
plt.xlabel(study_hours)
plt.ylabel(marks)
plt.show()
```

## 5. Pie chart

```python
import matplotlib.pyplot as plt

# Data values for each category
data = [60, 30, 10]

# Names of the categories
labels = ["AI", "ML", "Other"]

# Create a pie chart and show percentage
plt.pie(data, labels=labels, autopct="%1.1f%%")

# Add title to the chart
plt.title("Data Distribution")

# Display the chart
plt.show()
```

## 6. Line chart

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 120, 90, 150]

plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```
