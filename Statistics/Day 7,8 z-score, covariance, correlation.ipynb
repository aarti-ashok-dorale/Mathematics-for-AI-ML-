# z-score

```python
import numpy as np
from scipy.stats import zscore

data = [10, 20, 30, 40, 50]

z_scores = zscore(data)

print(z_scores) 
```

# Covariance

```python
x = [2,4,6,8]
y = [40,50,60,70]

print(np.cov(x,y)[0,1])
```

# Correlation

```python
import numpy as np

x = [2, 4, 6, 8]
y = [40, 50, 60, 70]

correlation = np.corrcoef(x, y)[0, 1]

print(correlation)
```
