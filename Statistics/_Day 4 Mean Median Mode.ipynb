## Statistics for AIML

### Mean = Average

```python
data = [10,20,30,30,40,50]
mean = sum(data)/len(data)
print(mean)
```

```python
# using numpy
import numpy as np
print(np.mean(data))
```

```python
# using pandas
import pandas as pd
print(pd.Series(data).mean()) 
```

### Median = Middle value

```python
# using numpy
data1 = [1,2,3,4,5]
print(np.median(data1))
```

```python
print(np.median(data))
```

```python
# using pandas
print(pd.Series(data).median())
```

```python
print(pd.Series(data1).median())
```

### Mode = Most frequently occuring values 

```python
# using numpy 
print(pd.Series(data).mode())
```

```python
# using statictics module
from statistics import mode
print(mode(data))
```
