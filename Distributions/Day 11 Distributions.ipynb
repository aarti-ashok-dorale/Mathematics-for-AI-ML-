# Distributions

## 1. Normal Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

# Normal Distribution मधून 100 values तयार करणे
data = np.random.normal(50, 10, 100)

# Histogram
plt.hist(data, bins=10) # bins=10 → Divides the data into 10 groups/ranges : 10-15, 16-20, ........

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

## 2. Bernoulli Distributions : 1 trial +2 outcomes

```python
# 1 = Success, 0 = Failure
result = np.random.binomial(1,0.5,10) #1 → Number of trials = 1
                                      # 0.5 → Probability of success = 50%
                                       #10 → Number of results = 10
print(result)
```

## 3. Binomial Distributions : Many trials + Number of Success

```python
result = np.random.binomial(10,0.5,5)
print(result)
```

## 4.Poisson Distribution

```python
# Average = 5 events
result = np.random.poisson(5, 10)

print(result) 
```
